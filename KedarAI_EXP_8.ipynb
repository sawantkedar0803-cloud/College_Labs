# AI Lab Experiment 8: AI-based Student Eligibility System

**Objective:** To represent knowledge using First Order Logic (FOL) and implement rule-based reasoning for intelligent decision-making.

**Problem Statement:** Determine student eligibility for campus placement based on CGPA and Attendance criteria.

```python
class StudentEligibilitySystem:
    def __init__(self):
        # First Order Logic Rule representation:
        # ∀x (CGPA(x) ≥ 7 ∧ Attendance(x) ≥ 75) → Eligible(x)
        self.rules = "IF CGPA(x) >= 7 AND Attendance(x) >= 75 THEN Eligible(x)"

    def is_eligible(self, name, cgpa, attendance):
        has_cgpa = cgpa >= 7.0
        has_attendance = attendance >= 75.0

        if has_cgpa and has_attendance:
            return True, "Eligible"
        else:
            return False, "Not Eligible"

    def run_inference(self, students):
        print(f"\n{'Name':<15} | {'CGPA':<6} | {'Attendance':<12} | {'Status'}")
        print("-" * 55)

        for student in students:
            name, cgpa, attendance = student
            is_elig, status = self.is_eligible(name, cgpa, attendance)
            print(f"{name:<15} | {cgpa:<6} | {attendance:<10}% | {status}")

system = StudentEligibilitySystem()

print("=== AI-based Student Eligibility System (FOL & Rule-Based) ===")
print("1. Running with predefined sample student database:")

# Sample Student Data (Knowledge Base)
student_data = [
    ("Rahul", 8.2, 80),
    ("Priya", 6.5, 60),
    ("Amit", 7.0, 75),
    ("Sneha", 9.1, 70),
    ("Vikram", 6.8, 90)
]

system.run_inference(student_data)

print("\n--- Interactive Student Evaluation ---")
interactive_students = list(student_data)

while True:
    choice = input("Do you want to check another student? (y/n): ").strip().lower()
    if choice != 'y':
        break

    name = input("Enter Student Name: ").strip()
    try:
        cgpa = float(input("Enter CGPA (0-10): "))
        attendance = float(input("Enter Attendance (%): "))
        interactive_students.append((name, cgpa, attendance))
    except ValueError:
        print("Invalid input! Please enter numeric values for CGPA and Attendance.")

if len(interactive_students) > len(student_data):
    print("\nUpdated Evaluation Including Your Inputs:")
    system.run_inference(interactive_students)
```

```python
class LoanApprovalSystem:
    def __init__(self):
        # First Order Logic Rule representation:
        # ∀x (CIBIL(x) ≥ 750 ∧ Income(x) ≥ 50000 ∧ ExistingLoans(x) ≤ 2) → LoanApproved(x)
        self.rules = "IF CIBIL >= 750 AND Income >= 50000 AND ExistingLoans <= 2 THEN Loan Approved ELSE Loan Rejected"

    def evaluate_loan(self, name, cibil, income, existing_loans):
        """
        Inference Engine checking conditions against the rules.
        """
        has_good_cibil = cibil >= 750
        has_good_income = income >= 50000
        has_few_loans = existing_loans <= 2

        if has_good_cibil and has_good_income and has_few_loans:
            return True, "Approved"
        else:
            reasons = []
            if not has_good_cibil:
                reasons.append("Low CIBIL (< 750)")
            if not has_good_income:
                reasons.append("Low Income (< 50,000)")
            if not has_few_loans:
                reasons.append("Too Many Existing Loans (> 2)")
            return False, f"Rejected ({', '.join(reasons)})"

    def run_inference(self, applicants):
        print(f"\n{'Applicant Name':<15} | {'CIBIL':<6} | {'Income (₹)':<12} | {'Loans':<6} | {'Status'}")
        print("-" * 65)

        for applicant in applicants:
            name, cibil, income, loans = applicant
            is_approved, status = self.evaluate_loan(name, cibil, income, loans)
            print(f"{name:<15} | {cibil:<6} | {income:<12} | {loans:<6} | {status}")

system = LoanApprovalSystem()

print("=== AI-based Loan Approval Rule-Based System (FOL & Inference Engine) ===")
print("1. Running with predefined sample applicant database:")

applicant_data = [
    ("Rahul Sharma", 780, 65000, 1),
    ("Priya Verma", 680, 55000, 0),
    ("Amit Patel", 810, 45000, 1),
    ("Sneha Gupta", 760, 90000, 3),
    ("Vikram Singh", 750, 50000, 2)
]

system.run_inference(applicant_data)

print("\n--- Interactive Loan Application Evaluation ---")
interactive_applicants = list(applicant_data)

while True:
    choice = input("Do you want to evaluate another loan application? (y/n): ").strip().lower()
    if choice != 'y':
        break

    name = input("Enter Applicant Name: ").strip()
    try:
        cibil = int(input("Enter CIBIL Score (300-900): "))
        income = float(input("Enter Monthly Income (₹): "))
        existing_loans = int(input("Enter Number of Existing Loans: "))
        interactive_applicants.append((name, cibil, income, existing_loans))
    except ValueError:
        print("Invalid input! Please enter valid numeric values for CIBIL score, income, and existing loans.")

if len(interactive_applicants) > len(applicant_data):
    print("\nUpdated Loan Evaluation Including Your Inputs:")
    system.run_inference(interactive_applicants)
```
